```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-zA-Z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df_clean.columns.tolist())
```

```python
print("Missing values per column:")
print(df_clean.isnull().sum())

print("\nExact duplicate rows:", df_clean.duplicated().sum())
```

```python
df_clean.info()
```

```python
# Convert date columns
for col in ["order_date", "ship_date"]:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(
            df_clean[col], errors="coerce"
        )

# Convert numeric columns if they exist
for col in ["sales", "profit", "discount", "quantity"]:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(
            df_clean[col], errors="coerce"
        )

print(df_clean[[
    col for col in ["order_date", "ship_date", "sales",
                    "profit", "discount", "quantity"]
    if col in df_clean.columns
]].dtypes)
```

```python
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))

print("\nMissing values after conversion:")
print(df_clean.isnull().sum())

print("\nExact duplicate rows:", df_clean.duplicated().sum())
```

```python
# Calculate key metrics(Business KPIs) for the retail sales dataset
total_sales = df_clean["sales"].sum()
total_profit = df_clean["profit"].sum()
total_orders = df_clean["order_id"].nunique()
total_customers = df_clean["customer_id"].nunique()
total_products = df_clean["product_id"].nunique()

profit_margin = (total_profit / total_sales) * 100

print("TOTAL SALES:", round(total_sales, 2))
print("TOTAL PROFIT:", round(total_profit, 2))
print("TOTAL ORDERS:", total_orders)
print("TOTAL CUSTOMERS:", total_customers)
print("UNIQUE PRODUCTS:", total_products)
print("PROFIT MARGIN:", round(profit_margin, 2), "%")
```

```python
#This helps identify which product categories contribute the most sales and profit
# Analyze sales and profit by category
category_analysis = (
    df_clean.groupby("category")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_quantity=("quantity", "sum")
    )
    .sort_values("total_sales", ascending=False)
)

category_analysis
```

```python
#Analyze reginonl performance by calculating total sales and profit for each region
#This helps identify the strongest and weakest regions by profit.
region_analysis = (
    df_clean.groupby("region")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_orders=("order_id", "nunique")
    )
    .sort_values("total_profit", ascending=False)
)

region_analysis
```

```python
#Analyze montly sales and profit trends to identify seasonal patterns and trends over time
df_clean["year_month"] = (
    df_clean["order_date"].dt.to_period("M")
)

monthly_analysis = (
    df_clean.groupby("year_month")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    )
)

monthly_analysis
```

```python
print("TOTAL SALES:", total_sales)
print("TOTAL PROFIT:", total_profit)
print("TOTAL ORDERS:", total_orders)
print("TOTAL CUSTOMERS:", total_customers)
print("TOTAL PRODUCTS:", total_products)
print("PROFIT MARGIN:", round(profit_margin, 2), "%")

display(category_analysis)
display(region_analysis)
display(monthly_analysis.head())
```

```python
#Profitability by sub-category
#Business question: Which products or product groups generate the highest and lowest profit?
# Group the dataset by product sub-category
subcategory_analysis = (
    df_clean.groupby("sub_category")

    # Calculate total sales, total profit and total quantity
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_quantity=("quantity", "sum")
    )

    # Sort sub-categories from lowest to highest profit
    .sort_values("total_profit", ascending=True)
)

# Display the complete sub-category analysis
display(subcategory_analysis)
```

```python
#Discount impact on profit
#Business question: Are higher discounts associated with lower profits?
# Group records by discount percentage
discount_analysis = (
    df_clean.groupby("discount")

    # Calculate total sales, profit and average profit
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        average_profit=("profit", "mean")
    )

    # Sort discount levels from lowest to highest
    .sort_index()
)

# Display the discount analysis
display(discount_analysis)
```

```python
"""Identify loss-making products
Business question: Which individual products have generated an overall loss?"""
# Group all transactions by product name
product_profit_analysis = (
    df_clean.groupby("product_name")

    # Calculate total sales and total profit for each product
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum")
    )
)

# Keep only products whose overall profit is negative
loss_making_products = product_profit_analysis[
    product_profit_analysis["total_profit"] < 0
]

# Sort products from the biggest loss to the smallest loss
loss_making_products = loss_making_products.sort_values(
    by="total_profit",
    ascending=True
)

# Display the 10 products with the largest losses
display(loss_making_products.head(10))
```

# Retail Sales & Profitability Analytics — Key Findings

## 1. Category Performance
- Technology is the highest-performing category by both sales and profit.
- Furniture generates substantial sales but comparatively low profit.
- Office Supplies generates strong profit despite lower sales.

## 2. Sub-category Performance
- Copiers are the most profitable sub-category.
- Phones and Accessories are strong profit contributors.
- Tables have the largest sub-category loss, followed by Bookcases and Supplies.

## 3. Discount Analysis
- Transactions with no discount generate the highest total profit.
- The 20% discount group also generates substantial profit.
- All discount levels from 30% to 80% show negative aggregate profit in this dataset.
- Further analysis is required to isolate the effect of discounts from product and regional differences.

## 4. Product-Level Analysis
- Cubify CubeX Double Head Printer is the largest loss-making product.
- Several printer, table and office-equipment products appear among the top 10 loss-makers.
- These products require further investigation of pricing, discounting and shipping costs.

## 5. Business Recommendations
1. Review discount policies, particularly at higher discount levels.
2. Investigate loss-making products and sub-categories.
3. Focus on high-profit products such as Copiers, Phones and Accessories.
4. Evaluate Furniture pricing and cost structure.
5. Monitor profit alongside sales when evaluating business performance.

```python
# Create a copy of the existing sub-category analysis
# This prevents us from modifying the original analysis table
subcategory_analysis = subcategory_analysis.copy()

# Calculate profit margin as a percentage of sales
# Multiply by 100 to convert the ratio into a percentage
subcategory_analysis["profit_margin"] = (
    subcategory_analysis["total_profit"]
    / subcategory_analysis["total_sales"]
) * 100

# Round the profit margin to two decimal places for readability
subcategory_analysis["profit_margin"] = (
    subcategory_analysis["profit_margin"].round(2)
)

# Sort sub-categories from highest to lowest profit margin
subcategory_analysis = subcategory_analysis.sort_values(
    by="profit_margin",
    ascending=False
)

# Display sales, profit and profit margin together
display(subcategory_analysis)
```
